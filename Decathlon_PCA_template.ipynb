# Principal component analysis

We can perform a Principal Component Analysis (PCA) to look at how all of the events go together.

This lets us answer questions such as:
- Which sports should you be good at if you want to win decathlon?
- Which sports do competitors tend to be good at in groups?

# Plot the PCA of competitors coloured by the competition they took part in

Extended activity: Did the competitors perform similarly in both competitions? Add back the Competition column to the scaled results.

Plot the PCA of individuals coloured by the competition.
Plot the PCA loadings labelled by each event.



```r
# add Competition back to the scaled data
dec_scaled$Competition <- decathlon$Competition
```

```r
# create a data frame of PC1, PC2, Competition and Athlete
pca_df <- data.frame(
  PC1 = res.pca$x[, 1],
  PC2 = res.pca$x[, 2],
  Competition = dec_scaled$Competition,
  Athlete = rownames(decathlon)
)

# set the colours for plotting
cols <- ifelse(pca_df$Competition == "Decastar", "tomato", "steelblue")

# Plot PC1 and PC2 labeled by Competition
plot(pca_df$PC1, pca_df$PC2,
     col = cols,
     pch = 19,
     xlab = "PC1",
     ylab = "PC2",
     main = "PCA: individuals")
text(pca_df$PC1, pca_df$PC2, labels = pca_df$Athlete, pos = 3, cex = 0.6)
legend("topright",
       legend = c("Decastar", "OlympicG"),
       col = c("tomato", "steelblue"),
       pch = 19,
       bty = "n")

```

```r
# create a dataframe with the PCA loadings
loadings <- as.data.frame(res.pca$rotation[, 1:2])
loadings$event <- rownames(loadings)

# plot the loadings and label by event
plot(loadings$PC1, loadings$PC2,
     type = "n",
     xlab = "PC1",
     ylab = "PC2",
     main = "PCA: variables")
abline(h = 0, v = 0, lty = 3)
arrows(0, 0, loadings$PC1, loadings$PC2, length = 0.08)
text(loadings$PC1, loadings$PC2, labels = loadings$event, pos = 3, cex = 0.7)
```

# Interpretation

We can see that Rank and Points are pointing in opposite directions along the x axis (first principal component). This makes sense because the more points you get, the higher ranked you will be.

We see that 100m and 100m hurdle correlates with rank. This means that being good at these events is more likely to give you a good rank. We see that Long Jump and High Jump are correlated with Points. This means that being good at these events is more likely to give you a high number of points. Bear in mind that a good score for 100m is lower whereas a good score for long jump is higher!
